```python
#CODE FOR WIND DEMAND PROFILE


import requests
import pandas as pd

# --- CONFIGURATION ---
TOKEN = '5deccdb8e80ef93a5458996d8e4635b1cc5faed3'
LAT = 52.356   # Science Park
LON = 4.950
YEAR = '2024'

# --- API REQUEST FOR WIND ---
api_base = 'https://www.renewables.ninja/api/data/wind'
headers = {'Authorization': 'Token ' + TOKEN}
params = {
    'lat': LAT, 'lon': LON,
    'date_from': f'{YEAR}-01-01', 'date_to': f'{YEAR}-12-31',
    'dataset': 'merra2',
    'capacity': 1.0,           # 1kW unit profile
    'height': 100,             # Hub height in meters
    'turbine': 'Vestas V80 2000', # Standard turbine model
    'format': 'json'
}

print("Fetching 2024 Wind Data for Zip 1098...")
response = requests.get(api_base, params=params, headers=headers)

if response.status_code == 200:
    data = response.json()
    df = pd.DataFrame.from_dict(data['data']).transpose()
    
    # Using 'unit=ms' again for the timestamps
    df.index = pd.to_datetime(df.index, unit='ms')
    
    filename = f'wind_unit_profile_1098_{YEAR}.csv'
    df.to_csv(filename)
    print(f"SUCCESS! Wind file saved as: {filename}")
else:
    print(f"Error: {response.status_code}. {response.text}")
```
